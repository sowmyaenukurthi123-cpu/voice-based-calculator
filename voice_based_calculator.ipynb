```python

from google.colab import output
import re

print("====================================")
print("       VOICE BASED CALCULATOR")
print("====================================")
print("Say a calculation like:")
print("Five plus three")
print("Ten minus two")
print("Six multiplied by four")
print("Twenty divided by five")

# Voice recognition using browser microphone
javascript = """
async function recognizeSpeech() {
    return new Promise((resolve, reject) => {
        const SpeechRecognition =
            window.SpeechRecognition || window.webkitSpeechRecognition;

        if (!SpeechRecognition) {
            reject("Speech recognition is not supported in this browser.");
            return;
        }

        const recognition = new SpeechRecognition();
        recognition.lang = 'en-US';
        recognition.continuous = false;
        recognition.interimResults = false;

        recognition.onresult = (event) => {
            resolve(event.results[0][0].transcript);
        };

        recognition.onerror = (event) => {
            reject(event.error);
        };

        recognition.start();
    });
}
recognizeSpeech();
"""

try:
    print("\nListening... Speak your calculation.")
    voice_text = output.eval_js(javascript)

    print("Recognized Voice:", voice_text)

    # Convert spoken numbers into digits
    numbers = {
        "zero": "0", "one": "1", "two": "2",
        "three": "3", "four": "4", "five": "5",
        "six": "6", "seven": "7", "eight": "8",
        "nine": "9", "ten": "10",
        "eleven": "11", "twelve": "12",
        "thirteen": "13", "fourteen": "14",
        "fifteen": "15", "sixteen": "16",
        "seventeen": "17", "eighteen": "18",
        "nineteen": "19", "twenty": "20"
    }

    text = voice_text.lower()

    for word, digit in numbers.items():
        text = re.sub(r'\b' + word + r'\b', digit, text)

    # Identify mathematical operator
    operators = {
        "plus": "+",
        "add": "+",
        "minus": "-",
        "subtract": "-",
        "multiplied by": "*",
        "times": "*",
        "multiply": "*",
        "divided by": "/",
        "divide": "/"
    }

    for word, symbol in operators.items():
        text = text.replace(word, symbol)

    # Extract calculation
    match = re.search(r'(\d+(?:\.\d+)?)\s*([+\-*/])\s*(\d+(?:\.\d+)?)', text)

    if match:
        a = float(match.group(1))
        operator = match.group(2)
        b = float(match.group(3))

        if operator == "+":
            result = a + b
        elif operator == "-":
            result = a - b
        elif operator == "*":
            result = a * b
        elif operator == "/":
            if b == 0:
                raise ZeroDivisionError("Cannot divide by zero")
            result = a / b

        print("\n====================================")
        print("             RESULT")
        print("====================================")
        print("Calculation:", a, operator, b)
        print("Answer:", result)
        print("====================================")

    else:
        print("Sorry! Could not understand the calculation.")

except Exception as e:
    print("Error:", e)
    print("Please allow microphone access and try again.")
```
